```python
# 1. Create an empty dictionary
inventory = {}
```

```python
# 2. Store the first product details in variables
ProductNo1_Name = "Mobile Phone"
ProductNo1_Quantity = 5
ProductNo1_Price = 20000
ProductNo1_releaseYear = 2020
```

```python
# 3. Add first product details to inventory
inventory["ProductNo1_Name"] = ProductNo1_Name
inventory["ProductNo1_Quantity"] = ProductNo1_Quantity
inventory["ProductNo1_Price"] = ProductNo1_Price
inventory["ProductNo1_releaseYear"] = ProductNo1_releaseYear
```

```python
# 4. Store the second product details in variables
ProductNo2_Name = "Laptop"
ProductNo2_Quantity = 3
ProductNo2_Price = 50000
ProductNo2_releaseYear = 2023
```

```python
# 5. Add second product details to inventory
inventory["ProductNo2_Name"] = ProductNo2_Name
inventory["ProductNo2_Quantity"] = ProductNo2_Quantity
inventory["ProductNo2_Price"] = ProductNo2_Price
inventory["ProductNo2_releaseYear"] = ProductNo2_releaseYear
```

```python
# 6. Display the products present in inventory
print("Inventory:")
print(inventory)
```

```python
# 7. Check if release year keys are in inventory
print("Product 1 Release Year exists:",
      "ProductNo1_releaseYear" in inventory)

print("Product 2 Release Year exists:",
      "ProductNo2_releaseYear" in inventory)
```

```python
# 8. Delete release year of both products
del inventory["ProductNo1_releaseYear"]
del inventory["ProductNo2_releaseYear"]

print("\nInventory after deleting release years:")
print(inventory)
```
