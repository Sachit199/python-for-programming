```python
# 1. Create an empty list
shopping_list = []
```

```python
# 2. Store the items in the shopping list
shopping_list = ["Watch", "Laptop", "Shoes", "Pen", "Clothes"]
```

```python
# 3. Add Football
shopping_list.append("Football")

```

```python
# 4. Print the last item
print("Last item:", shopping_list[-1])


```

```python
# 5. Print the entire shopping list
print("Shopping list:", shopping_list)
```

```python
# 6. Print important items: Laptop and Shoes
print("Important items:", shopping_list[1:3])
```

```python
# 7. Replace Pen with Notebook
shopping_list[3] = "Notebook"
print("After replacing Pen:", shopping_list)
```

```python
# 8. Delete the least important item (Football)
shopping_list.remove("Football")
print("Final shopping list:", shopping_list)
```
