```python
import pandas as pd

customer_data = {
    "Customer_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Name": ["Rahul", "Priya", "Aman", "Sneha", "Arjun",
             "Neha", "Rohan", "Anjali", "Vikram", "Kavya"],
    "Age": [25, 34, 29, 41, 36, 22, 45, 31, 28, 39],
    "Gender": ["Male", "Female", "Male", "Female", "Male",
              "Female", "Male", "Female", "Male", "Female"],
    "Purchase_Amount": [2500, 4200, 1800, 5500, 3200,
                        1500, 6000, 2800, 3700, 4500]
}

customers = pd.DataFrame(customer_data)

customers
```
