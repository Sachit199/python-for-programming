```python
#Q1. Check Lionel Messi's achievements
name = "Lionel Messi"
sport = "Soccer"
achievements = 7

if achievements > 10:
    print(name, sport, achievements)
else:
    print(name, "does not have more than 10 achievements")
```

```python
#Q2. Check Tennis or exactly 20 achievements
name = "Roger Federer"
sport = "Tennis"
achievements = 20

if sport == "Tennis" or achievements == 20:
    print("Success")
else:
    print("Condition not satisfied")
```

```python
#Q3. Check less than 10 achievements and does not play Soccer
name = "Usain Bolt"
sport = "Athletics"
achievements = 8

if achievements < 10 and sport != "Soccer":
    print(name, sport, achievements)
else:
    print("Criteria not met")
```

```python
#Q4. Integer input (1 to 999)
num = int(input("Enter an integer from 1 to 999: "))

# 1. Even or Odd
if num % 2 == 0:
    print("Even")
else:
    print("Odd")

# 2. Palindrome
if str(num) == str(num)[::-1]:
    print("Palindrome")
else:
    print("Not a palindrome")

# 3. Divisible by sum of digits
digit_sum = sum(int(digit) for digit in str(num))

if num % digit_sum == 0:
    print("Divisible by the sum of its digits")
else:
    print("Not divisible by the sum of its digits")
```

```python
#Q6. Print the list using a for loop
squares = ['red', 'yellow', 'green', 'purple', 'blue']

for color in squares:
    print(color)
```

```python
#Q7. Copy values until 'orange'
squares = ['red', 'yellow', 'green', 'purple', 'blue', 'orange']

new_squares = []

for color in squares:
    if color == 'orange':
        break

    new_squares.append(color)

print(new_squares)
```

```python
#Q8. Find animals with exactly 7 letters
Animals = ['lion', 'giraffe', 'gorilla', 'parrots', 'crocodile', 'deer', 'swan']

seven_letter_animals = []

i = 0

while i < len(Animals):
    if len(Animals[i]) == 7:
        seven_letter_animals.append(Animals[i])

    i += 1

print(seven_letter_animals)

```
